```python
import pandas as pd
```

```python
#df = pd.read_csv("LA_O3_1980_2005.csv")

df = pd.read_csv("LA_O3_2005_2024.csv")
df.head()
```

```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
df['year'] = df.Date.dt.year
```

```python
df.columns
```

```python
df_1 = df.groupby('year')[['Daily Max 8-hour Ozone Concentration']].mean().reset_index()#.to_csv('LA_annualavg.csv')
```

```python
df_2 = df.groupby('year')[['Daily Max 8-hour Ozone Concentration']].mean().reset_index()#.to_csv('LA_annualavg.csv')
```

```python
pd.concat([df_1,df_2]).to_csv('LA_O3_annualavg.csv', index=False)
```
